## Timestamp-only preview (privacy-preserving)
Use the real publisher CSV/ZIP path to see **only three original Timestamp values**. Never display source/destination IPs or original data rows.


```python
# Select ONE real publisher CSV or ZIP after inspecting the mounted file list.
PREVIEW_DATA = None  # e.g. Path('/kaggle/input/<publisher-archive>/GeneratedLabelledFlows.zip')
if PREVIEW_DATA is None:
    print('Attach and identify the publisher file before previewing timestamps.')
else:
    import zipfile, pandas as pd
    PREVIEW_DATA=Path(PREVIEW_DATA)
    if PREVIEW_DATA.suffix.lower()=='.zip':
        with zipfile.ZipFile(PREVIEW_DATA) as z:
            candidates=sorted(n for n in z.namelist() if n.lower().endswith('.csv'))
            assert candidates,'No CSV files found inside publisher ZIP'
            with z.open(candidates[0]) as stream:
                tiny=pd.read_csv(stream,nrows=3,encoding='utf-8-sig')
            print('Previewed CSV member:',candidates[0])
    else:
        tiny=pd.read_csv(PREVIEW_DATA,nrows=3,encoding='utf-8-sig')
    from importlib.util import spec_from_file_location, module_from_spec
    module_spec=spec_from_file_location('e7_g0_source',REPO/'scripts/e7_g0_external_source_audit.py')
    module=module_from_spec(module_spec)
    module_spec.loader.exec_module(module)
    mapping=module.find_columns(tiny.columns)
    print('Available canonical fields:',sorted(mapping))
    if 'time' in mapping:
        print('Timestamp strings ONLY:',tiny[mapping['time']].astype(str).tolist())
    else:
        print('Timestamp column absent: G0 cannot pass. No raw rows/IPs displayed.')
```
